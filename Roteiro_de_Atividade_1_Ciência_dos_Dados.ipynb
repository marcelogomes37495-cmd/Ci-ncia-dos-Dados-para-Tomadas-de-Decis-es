### Resposta 3
### Interpretação dos tipos de características

Embora a maioria das colunas apareça tecnicamente como `int64`, nem todas representam variáveis numéricas contínuas.

As variáveis `LIMIT_BAL`, `AGE`, `BILL_AMT1` até `BILL_AMT6` e `PAY_AMT1` até `PAY_AMT6` são consideradas **numéricas**, pois representam valores quantitativos.

As variáveis `SEX`, `EDUCATION`, `MARRIAGE`, `PAY_1` até `PAY_6` e `default payment next month` são consideradas **categóricas**, pois os números representam classes ou categorias discretas.

A variável `ID` é classificada como **metadado**, pois serve apenas como identificador único de cada registro.


```python
# Pergunta 4 — Aparência dos dados segundo as características

print("Resumo estatístico das variáveis numéricas:")
print(df.describe())

print("\nFrequência das variáveis categóricas:")

categoricas = ['SEX', 'EDUCATION', 'MARRIAGE', 'PAY_1', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6', 'default payment next month']

for col in categoricas:
    print(f"\nFrequência de {col}:")
    print(df[col].value_counts())
```

### Resposta 4
### Análise da aparência dos dados

A análise estatística das variáveis numéricas mostrou grande variação entre os valores observados no conjunto de dados. A variável `LIMIT_BAL`, por exemplo, apresenta ampla diferença entre os limites mínimos e máximos de crédito concedido.

As variáveis de fatura (`BILL_AMT`) e pagamento (`PAY_AMT`) também apresentam dispersão elevada, indicando diferentes perfis financeiros entre os clientes.

Nas variáveis categóricas, observou-se predominância de determinadas classes em `SEX`, `EDUCATION` e `MARRIAGE`.

As variáveis `PAY_1` até `PAY_6` mostram diferentes frequências de atraso nos pagamentos, permitindo identificar padrões de comportamento financeiro.

Na variável `default payment next month`, a maioria dos registros corresponde a clientes sem inadimplência no mês seguinte.
