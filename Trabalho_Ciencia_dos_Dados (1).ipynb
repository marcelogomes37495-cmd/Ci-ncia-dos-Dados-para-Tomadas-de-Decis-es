```python
df = pd.read_excel('default_of_credit_card_clients__courseware_version_1_21_19.xls')
```

```python
# Pergunta 1 — Quantas colunas?
print("Número de colunas:", df.shape[1])
print("Nomes:", df.columns.tolist())
```

```python
# Pergunta 2 — Quantas linhas (amostras)?
print("Número de linhas:", df.shape[0])
```

```python
# Pergunta 3 — Tipos de características
print(df.dtypes)
print("\nNuméricas:", df.select_dtypes(include='number').columns.tolist())
print("Não-numéricas:", df.select_dtypes(exclude='number').columns.tolist())
```

```python
# Pergunta 4 — Aparência dos dados
df.describe()   # Resumo estatístico das colunas numéricas
```

```python
# Pergunta 5 — Dados faltando?
print("Valores nulos por coluna:")
print(df.isnull().sum())
print("\nTotal de valores nulos:", df.isnull().sum().sum())
```
